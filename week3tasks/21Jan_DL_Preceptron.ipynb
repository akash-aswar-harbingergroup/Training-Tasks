```python
# Scenario: Employee Promotion Decision (Single-Layer Perceptron)
# Inputs (2 features):
# Years of experience (0 = low, 1 = sufficient)
# Performance rating (0 = poor, 1 = good)
# Output:
# 1 = gets promoted
# 0 = not promoted
# Rule (intuitive):
# Employee gets promoted only if both experience and performance are good.
# This is again an AND logic gate, but in a workplace context.
 
```

```python
import numpy as np

# Data: unique AND gate cases
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1],
])
y = np.array([0, 0, 0, 1])  # AND truth table
```

```python
# Add bias term by augmenting X with a column of 1s
X_aug = np.hstack([X, np.ones((X.shape[0], 1))])

np.random.seed(0)
w = np.random.randn(3) * 0.1  # weights: [w1, w2, b]

lr = 0.1
epochs = 25

def step(z): 
    return (z >= 0).astype(int)
```

```python
for epoch in range(epochs):
    total_errors = 0
    for xi, target in zip(X_aug, y):
        z = np.dot(xi, w)
        y_hat = step(z)
        error = target - y_hat
        if error != 0:
            w += lr * error * xi  # perceptron update
            total_errors += 1
    # Optional: print progress
    # print(f"Epoch {epoch+1:02d} | Misclassified: {total_errors}")

print("Final weights (w1, w2, b):", w)
```

```python
# Predictions
print("\nPredictions:")
for xi, target in zip(X_aug, y):
    z = np.dot(xi, w)
    y_hat = step(z)
    print(f"Input {xi[:2]} -> Pred: {y_hat} (Target: {target})")
```
