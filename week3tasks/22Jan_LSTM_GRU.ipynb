```python
# A smart city power department wants to predict tomorrow’s 
# electricity demand (in MW) using the last 7 days of hourly consumption data.

# This helps to:
# Avoid power shortages ⚡
# Optimize power generation scheduling
# Reduce operational costs
# 🔹 Sample Dataset (Simplified – Daily Average)
```

```python

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, GRU, Dense
from tensorflow.keras.callbacks import EarlyStopping
from sklearn.model_selection import train_test_split
```

```python
demand = np.array([320, 340, 360, 355, 370, 390, 410, 430, 450, 470], dtype=float)

window_size = 7   # use last 7 days to predict next day

# Normalization with mean/std
mean = demand.mean()
std = demand.std()
demand_norm = (demand - mean) / std
```

```python
X, y = [], []

for i in range(len(demand_norm) - window_size):
    X.append(demand_norm[i:i + window_size])
    y.append(demand_norm[i + window_size])

X = np.array(X)
y = np.array(y)

# reshape for RNN input
X = X.reshape((X.shape[0], window_size, 1))

# Train-test split (80/20)
X_train, X_val, y_train, y_val = train_test_split(X, y, test_size=0.2, shuffle=False)
```

```python
X, y = [], []

for i in range(len(demand_norm) - window_size):
    X.append(demand_norm[i:i + window_size])
    y.append(demand_norm[i + window_size])

X = np.array(X)
y = np.array(y)

# reshape for RNN input (samples, time_steps, features)
X = X.reshape((X.shape[0], window_size, 1))
```

```python
lstm_model = Sequential([
    LSTM(32, activation='tanh', input_shape=(window_size, 1)),
    Dense(1)
])

lstm_model.compile(optimizer='adam', loss='mse')

history_lstm = lstm_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=200,
    verbose=1
)
```

```python
gru_model = Sequential([
    GRU(32, activation='tanh', input_shape=(window_size, 1)),
    Dense(1)
])

gru_model.compile(optimizer='adam', loss='mse')

history_gru = gru_model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=200,
    verbose=1
)
```

```python
print("\n--- TRAINING LOSSES ---")
print("LSTM final train loss:", history_lstm.history['loss'][-1])
print("LSTM final val loss  :", history_lstm.history['val_loss'][-1])

print("GRU  final train loss:", history_gru.history['loss'][-1])
print("GRU  final val loss  :", history_gru.history['val_loss'][-1])
```

```python
# last 7 days of actual electricity demand
last_week = demand[-window_size:]
last_week_norm = ((last_week - mean) / std).reshape(1, window_size, 1)

# model predictions
pred_lstm_norm = lstm_model.predict(last_week_norm, verbose=0)[0, 0]
pred_gru_norm  = gru_model.predict(last_week_norm,  verbose=0)[0, 0]

# unnormalize predictions (back to MW)
pred_lstm = pred_lstm_norm * std + mean
pred_gru  = pred_gru_norm  * std + mean
```

```python
print("\n--- PREDICTION ---")
print("Last 7 days electricity demand (MW):", np.round(last_week, 1))
print(f"LSTM predicted next day electricity demand: {pred_lstm:.1f} MW")
print(f"GRU  predicted next day electricity demand: {pred_gru:.1f} MW")
```
